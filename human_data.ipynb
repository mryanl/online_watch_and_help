```python
import pickle
import json
from pathlib import Path

data_dir = Path("dataset")
data_file = data_dir / "test-2_per_apt-task_no_tv-apts_3,6.pik"

with open(data_file, "rb") as f:
    data = pickle.load(f)
```

```python
human_episodes_id = [1, 5, 8, 12]
human_episodes = []
human_dataset = data_dir / "human_data"
human_dataset.mkdir(exist_ok=True)
for ep in human_episodes_id:
    print(f"Episode {ep}:")
    print(data[ep])
    print("\n")
    human_episodes.append(data[ep])

    with open(human_dataset / f"{ep}.json", "w") as f:
        json.dump(data[ep], f, indent=4)


with open(human_dataset / "human_episodes.pik", "wb") as f:
    pickle.dump(human_episodes, f)
```
